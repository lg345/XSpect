```python
from XSpect import Pipeline
p = Pipeline.from_yaml('examples/mfxl1027922_ultrafast_xes.yaml')
p.run(cores=4, batch_size=500)
print(f'Runs processed: {len(p.analyzed_runs)}')
for i, run in enumerate(p.analyzed_runs):
    print(f'  Run {i}: time_binned shape = {run.epix_ROI_1_simultaneous_laser_time_binned.shape}')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Combine time-binned spectra across all runs (matches v.combine_spectra)
laser_on = np.zeros((40, 705))
laser_off = np.zeros((40, 705))
for run in p.analyzed_runs:
    laser_on += run.epix_ROI_1_simultaneous_laser_time_binned
    laser_off += run.epix_ROI_1_xray_not_laser_time_binned

# Use avg_all_laser_off approach for better SNR
# (matches XSpect_Visualization.combine_spectra with sub_mode='avg_all_laser_off')
# This averages all laser-off time bins into a single reference spectrum,
# eliminating the shot-count modulation noise in low-statistics laser-off bins
norm_laser_on = np.nansum(laser_on, axis=1)
laser_on_norm = np.divide(laser_on.T, norm_laser_on).T

laser_off_t_sum = np.nansum(laser_off, axis=0)
laser_off_tiled = np.tile(laser_off_t_sum, (laser_off.shape[0], 1))
norm_factor = np.divide(np.nansum(laser_on_norm), np.nansum(laser_off_tiled))
laser_off_norm = norm_factor * laser_off_tiled

diff = laser_on_norm - laser_off_norm
energy = p.analyzed_runs[0].kbeta_energy
time_bins = np.linspace(-0.9, 0.9, 40)

# SVD reconstruction (2 components, matches reference cell 10)
U, s, Vt = np.linalg.svd(diff[2:], full_matrices=False)
nsvd = 2
diff_svd = U[:, :nsvd] @ np.diag(s[:nsvd]) @ Vt[:nsvd, :]
plotmax = np.max(np.abs(diff_svd))
contlevels = np.linspace(-plotmax, plotmax, 50)

print(f"SV1 fraction: {s[0]**2/np.sum(s**2)*100:.1f}%")
print(f"Diff range: [{diff.min():.4f}, {diff.max():.4f}]")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: raw difference spectrum (matches v.plot_2d_difference_spectrum)
ax = axes[0]
im = ax.imshow(diff.T, cmap='RdBu', vmin=-0.006, vmax=0.004, origin='lower', aspect='auto')
ax.set_xlabel('Time bin')
ax.set_ylabel('Pixel')
ax.set_title('Raw difference (on - off)')
plt.colorbar(im, ax=ax)

# Right: SVD reconstruction (matches reference cell 10)
ax = axes[1]
cf = ax.contourf(energy, time_bins[2:], diff_svd, contlevels, cmap='RdBu')
ax.set_xlabel('Energy (eV)')
ax.set_ylabel('Time (ps)')
ax.set_title(f'SVD Reconstruction (n={nsvd})')
plt.colorbar(cf, ax=ax)

plt.tight_layout()
plt.show()

# Kinetic trace (matches reference cell 8)
fig2, ax2 = plt.subplots(figsize=(6, 4))
kin = np.nansum(np.abs(diff[2:, 340:368]), axis=1)
ax2.plot(time_bins[2:], kin, 'o-')
ax2.set_xlabel('Time (ps)')
ax2.set_ylabel(r'$\Delta$ XES')
ax2.set_title('Kinetic trace (pixels 340-368)')
plt.tight_layout()
plt.show()
```
